# Manejo de listas.

```python
Marcas=["Alfa Romeo", "Tsuru", "Shelby", "March", "Paganni", "Ferrari", "Corvette", "Porshe"]

# Tamaño de la lista.
print(len(Marcas))
len(Marcas)
```

```python
# Agregar un valor al final de la lista.
Marcas.append("Paganni")
Marcas
```

```python
# Agregar un nuevo valor en una posición especifica.
Marcas.insert(2, "Swift Sport")
Marcas
```

```python
# Buscar un elemento dentro de la lista.
print("Swift Sport" in Marcas)
print("MG 5" in Marcas)
```

```python
# Conocer la posición de un elemento.
print(Marcas.index("Swift Sport"))
```

```python
# Contar elementos repetidos.
Marcas.count("Paganni")
```

```python
# Eliminar el último elemento de la lista.
Marcas.pop()
```

```python
# Eliminar un elemento específico de la lista indicando la posición.
Marcas.pop(1)
```

```python
# Eliminar un elemento específico de la lista buscando por su nombre.
Marcas.remove("March")
Marcas
```

```python
# Invertir el orden de la lista.
Marcas.reverse()
Marcas
```

```python
# Sumar el contenido de 2 listas en 1.
Numbers = [15, 11, 13, 12, 14]

DosListas = Marcas + Numbers
DosListas
```

```python
# Multiplicar lista.
Numbers * 2
```

```python
# Ordenar elementos de la lista.
Numbers.sort()
Numbers
```

```python
# Invertir orden de la lista.
Numbers.sort(reverse=True)
Numbers
```
